## 2. First anchor: residue constraint

A first visible constraint appears modulo 6.

For every prime greater than 3:

\[
p > 3 \Rightarrow p \equiv \pm 1 \pmod{6}
\]

This does not identify every number congruent to \(1\) or \(5 \pmod{6}\) as prime.

It says something simpler and stronger for a first measurement:

> primes greater than 3 remain inside two residue classes modulo 6.

````python
# Minimal verification of the mod 6 anchor.

def simple_sieve(n_max: int):
    if n_max < 2:
        return []
    sieve = [True] * n_max
    sieve[0:2] = [False, False]
    p = 2
    while p * p < n_max:
        if sieve[p]:
            for k in range(p * p, n_max, p):
                sieve[k] = False
        p += 1
    return [i for i, is_prime in enumerate(sieve) if is_prime]

N_MAX = 100
primes = simple_sieve(N_MAX)

[(p, p % 6) for p in primes if p > 3][:20]
````

## 3. Core vocabulary

These terms will stay consistent across notebooks.

### Remains under constraint / persists

A structure **remains under constraint** when it is retained after filtering.

Short form for titles, captions, and summaries:

> **persists**

Example:

> Prime residues persist in \(\pm 1 \pmod{6}\).

Full form:

> Prime residues remain under constraint in \(\pm 1 \pmod{6}\).

### Drift

**Drift** is measurable deviation from constrained structure.

Drift is not used as a moral term and not as a vague synonym for failure.

It means:

> a measurement has moved away from the defined constraint, baseline, or recoverable structure.

### Recoverability

**Recoverability** asks whether structure can be reconstructed from partial or degraded information.

Example future question:

> If part of the prime sequence is hidden or corrupted, which residue, density, or gap features remain recoverable?

## 4. CGCS working definition

For this repo, CGCS is treated as a measurement habit:

> **Constraint-Gaged Comprehension Score:** a score or summary that asks how much structure remains under a stated constraint.

The score may vary by notebook.

Examples:

- residue notebook: fraction of primes in valid residue classes
- density notebook: agreement with \(x / \log x\)
- gap notebook: measurable structure in gap distributions
- recoverability notebook: reconstruction accuracy from partial observations

CGCS should stay measurable, reproducible, and notebook-specific.

## 6. What this repo does not claim

This repo does **not** claim to prove the Riemann Hypothesis.

This repo does **not** claim that residue constraints are enough to generate primes.

This repo does **not** claim that simple plots replace proof.

Instead, the repo starts from reproducible measurements and builds toward more advanced questions carefully.

## 7. Similarity and difference from zeta-constraint-lab

**prime-numbers-lab** studies direct prime structure:

- residues
- gaps
- density
- sieve layers
- finite recoverability

**zeta-constraint-lab** studies spectral / analytic structure:

- zeta zeros
- spacing statistics
- residual structure
- RH-adjacent tests
- reconstruction from spectral constraints

Compact bridge:

> finite constraints retain structure; spectral constraints encode it.
