## 3. Data generation or loading

For most early notebooks, generate primes directly. For larger or specialized notebooks, load cached data from `data/`.

```python
def generate_primes(n_max: int) -> np.ndarray:
    """Generate primes less than n_max.

    Uses SymPy when available; otherwise falls back to a simple sieve.
    """
    if n_max < 2:
        return np.array([], dtype=int)

    if HAS_SYMPY:
        return np.array(list(sp.primerange(2, n_max)), dtype=int)

    sieve = np.ones(n_max, dtype=bool)
    sieve[:2] = False
    for i in range(2, int(math.sqrt(n_max)) + 1):
        if sieve[i]:
            sieve[i*i:n_max:i] = False
    return np.nonzero(sieve)[0].astype(int)

primes = generate_primes(N_MAX)

summary = {
    "n_max": N_MAX,
    "prime_count": int(len(primes)),
    "first_primes": primes[:10].tolist(),
    "last_primes": primes[-10:].tolist() if len(primes) >= 10 else primes.tolist(),
}

summary
```

## 5. CGCS score

The CGCS score cell is optional but recommended.

Keep it measurable and notebook-specific.

Working definition:

> **CGCS score = fraction of target structure remaining under the stated constraint.**

Examples:

- residue notebook: fraction of primes in valid residues
- sieve notebook: retained candidates after valid constraint layers
- recoverability notebook: \(1 -\) normalized reconstruction error
- density notebook: bounded agreement score between observed and predicted density

```python
# Placeholder CGCS score.
# Replace with a notebook-specific score where possible.

# Example density agreement score:
# score approaches 1 when observed pi(x) is close to x/log(x).
relative_error = abs(pi_x - x_over_log_x) / max(pi_x, 1)
cgcs_score = max(0.0, 1.0 - relative_error)

cgcs = {
    "score": float(cgcs_score),
    "definition": "placeholder density agreement: 1 - |pi(x)-x/log(x)|/pi(x)",
    "interpretation": "Closer to 1 indicates stronger agreement under the selected constraint.",
}

cgcs
```
